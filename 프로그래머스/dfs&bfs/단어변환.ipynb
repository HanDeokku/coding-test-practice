```python
import sys
sys.setrecursionlimit(100000)

answer = 10000000

def dfs(words, change, target, visited, cnt):
    global answer
    if change == target:
        answer = min(answer, cnt)
        return
    
    if len(visited) == len(words):
        return
    
    for word in words:
        if word not in visited:
            count = 0
            for i in range(len(word)):
                if change[i] == word[i]:
                    count += 1
            if count == 1:
                visited.append(visited)
                dfs(words, word, target, visited, cnt + 1)
    
    

def solution(begin, target, words):
    global answer
    visited = []
    cnt = 0
    dfs(words, begin, target, visited, cnt)
    
    return answer
```
