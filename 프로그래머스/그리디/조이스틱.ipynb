```python
# 로직
# 조사 위치 확인 - 알파벳 체크
# 시작 위치 - 조사 위치
# 시작 위치 포인트 탐색
# 조사 위치 - 시작 위치 변경
def solution(name):
    answer = 0
    name_len = len(name)
    names = list(name)
    startNames = ['A'] * name_len
    
    survey = 0
    while ''.join(startNames) != ''.join(names):
        if startNames[survey] != names[survey]:
            if abs(ord(startNames[survey]) - ord(names[survey])) > 13:
                answer += (26 - abs(ord(startNames[survey]) - ord(names[survey])))
                startNames[survey] = names[survey]
            else:
                answer += abs(ord(startNames[survey]) - ord(names[survey]))
                startNames[survey] = names[survey]

        if ''.join(startNames) == ''.join(names):
            break

        score = [100] * name_len
        min_index = survey
        min_score = score[survey]
        for i in range(name_len):
            if (i != survey) and (startNames[i] != names[i]):
                score[i] = abs(i-survey)
                if score[i] > name_len//2:
                    score[i] = name_len - score[i]
                if min_score > score[i]:
                    min_index = i
                    min_score = score[i]
        survey = min_index
        answer += min_score
        
    return answer

name = "JAN"
print(solution(name))
```
