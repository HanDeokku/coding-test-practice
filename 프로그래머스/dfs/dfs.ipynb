```python
def dfs(graph, start):
    visited = set()  # 방문한 노드를 저장하는 집합
    stack = [start]  # 탐색할 노드를 저장하는 스택

    while stack:
        vertex = stack.pop()  # 스택에서 노드 하나를 꺼냄
        if vertex not in visited:
            print(vertex)  # 현재 노드 방문
            visited.add(vertex)  # 현재 노드를 방문한 것으로 표시
            stack.extend(neighbor for neighbor in graph[vertex] if neighbor not in visited)

# 예시 그래프
graph = {
    'A': ['B', 'C'],
    'B': ['A', 'D', 'E'],
    'C': ['A', 'F'],
    'D': ['B'],
    'E': ['B', 'F'],
    'F': ['C', 'E']
}

dfs(graph, 'A')
```
